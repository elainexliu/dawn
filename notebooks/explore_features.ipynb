# Feature exploration

Exploratory notebook — not production code.  Used to sanity-check feature
distributions, spot class imbalance, and visualize per-channel signals.

Run from repo root after generating `data/processed/features.npz` via the
pipeline (segmentation → features).

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

NPZ_PATH = Path("data/processed/features.npz")
```

```python
npz = np.load(NPZ_PATH, allow_pickle=True)
X      = npz["X"]
y      = npz["y"]
groups = npz["groups"]
feat_names = npz["feature_names"].tolist()

print(f"X shape:  {X.shape}")
print(f"Positive rate: {y.mean():.2%}")
print(f"Days: {np.unique(groups)}")
```

```python
df = pd.DataFrame(X, columns=feat_names)
df["label"] = y
df["day"]   = groups
df.head()
```

```python
# Compare feature distributions between classes
fig, axes = plt.subplots(4, 4, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(feat_names[:16]):
    ax = axes[i]
    for lbl, grp in df.groupby("label"):
        ax.hist(grp[col], bins=30, alpha=0.5, label=f"label={lbl}")
    ax.set_title(col, fontsize=8)
    ax.legend(fontsize=6)

plt.tight_layout()
plt.show()
```
